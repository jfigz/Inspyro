```python
# ============================================
# DISEÑO DE VIGA SIMPLEMENTE APOYADA
# ============================================
# Cálculo de esfuerzos y deflexión máxima
# para una viga con carga distribuida uniforme

print("=" * 50)
print("   DISEÑO DE VIGA SIMPLEMENTE APOYADA")
print("   Carga distribuida uniforme")
print("=" * 50)
```

```python
# --- Datos de entrada ---
L = 6.0       # Longitud de la viga [m]
w = 15.0      # Carga distribuida [kN/m]
b = 0.20      # Ancho de sección [m]
h = 0.40      # Altura de sección [m]
E = 200e6     # Módulo de elasticidad del acero [kN/m²]
fy = 250e3    # Esfuerzo de fluencia [kN/m²]

print(f"Longitud:           L  = {L} m")
print(f"Carga distribuida:  w  = {w} kN/m")
print(f"Sección:            {b*100:.0f} x {h*100:.0f} cm")
print(f"Módulo E:           {E/1e6:.0f} GPa")
print(f"Fluencia fy:        {fy/1e3:.0f} MPa")
```

```python
# --- Propiedades de la sección ---
I = (b * h**3) / 12     # Momento de inercia [m⁴]
S = (b * h**2) / 6      # Módulo de sección [m³]
A = b * h               # Área [m²]

print(f"Momento de inercia: I = {I*1e6:.4f} x 10⁻⁶ m⁴")
print(f"Módulo de sección:  S = {S*1e6:.4f} x 10⁻⁶ m³")
print(f"Área de sección:    A = {A*1e4:.2f} cm²")
```

```python
# --- Reacciones y esfuerzos ---
R = w * L / 2                    # Reacción en apoyos [kN]
V_max = R                        # Corte máximo [kN]
M_max = w * L**2 / 8             # Momento máximo [kN·m]

print("─" * 40)
print("  RESULTADOS DE ESFUERZOS")
print("─" * 40)
print(f"Reacción en apoyos:  R    = {R:.2f} kN")
print(f"Corte máximo:        Vmax = {V_max:.2f} kN")
print(f"Momento máximo:      Mmax = {M_max:.2f} kN·m")
```

```python
# --- Verificación de resistencia ---
sigma_max = M_max / S            # Esfuerzo de flexión máximo [kN/m²]
tau_max = (3 * V_max) / (2 * A)  # Esfuerzo cortante máximo [kN/m²]
ratio = sigma_max / fy           # Ratio demanda/capacidad

print("─" * 40)
print("  VERIFICACIÓN DE RESISTENCIA")
print("─" * 40)
print(f"Esfuerzo flexión:    σmax = {sigma_max/1e3:.2f} MPa")
print(f"Esfuerzo cortante:   τmax = {tau_max/1e3:.2f} MPa")
print(f"Ratio D/C:           {ratio:.3f}")
print(f"Estado:              {'✅ OK' if ratio < 1.0 else '❌ EXCEDE CAPACIDAD'}")
```

```python
# --- Deflexión ---
delta_max = (5 * w * L**4) / (384 * E * I)  # Deflexión máxima [m]
delta_lim = L / 360                          # Límite L/360 [m]

print("─" * 40)
print("  VERIFICACIÓN DE DEFLEXIÓN")
print("─" * 40)
print(f"Deflexión máxima:    δmax = {delta_max*1000:.3f} mm")
print(f"Deflexión límite:    δlim = {delta_lim*1000:.3f} mm  (L/360)")
print(f"Ratio:               {delta_max/delta_lim:.3f}")
print(f"Estado:              {'✅ OK' if delta_max < delta_lim else '❌ EXCEDE LÍMITE'}")
print()
print("=" * 50)
print("  CÁLCULO COMPLETADO EXITOSAMENTE")
print("=" * 50)
```
