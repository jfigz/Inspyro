# Example Beam Report

Minimal Inspyro demo for structural calculation and report authoring.

```python
import json
from pathlib import Path

from beam_design import (
    demand_capacity_ratio,
    factored_uniform_load,
    flexural_capacity_kNm,
    max_moment_kNm,
)

case = json.loads(Path('inputs/beam_case.json').read_text(encoding='utf-8'))
wu = factored_uniform_load(case['dead_load_kN_m'], case['live_load_kN_m'])
mu = max_moment_kNm(wu, case['span_m'])
phi_mn = flexural_capacity_kNm(case['steel_yield_MPa'], case['section_modulus_cm3'])
ratio = demand_capacity_ratio(mu, phi_mn)

{
    'project': case['project'],
    'factored_load_kN_m': round(wu, 2),
    'demand_kNm': round(mu, 2),
    'capacity_kNm': round(phi_mn, 2),
    'demand_capacity_ratio': round(ratio, 3),
    'status': 'OK' if ratio <= 1.0 else 'NG',
}
```
